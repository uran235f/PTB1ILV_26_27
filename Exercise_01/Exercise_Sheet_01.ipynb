## Problem 4

Write code to print the formatted multiplication table for a given number n.

### Pseudocode

```text
input n
i = 1
while i <= 10
    n = 1
    while n <= 10:
        result = n * i
        output n, i and result
        n = n + 1
    move to the next line
    i = i + 1

```

````python
i = 1
while i <= 10: 
    n = 1
    while n <= 10: 
        result = n * i
        print("{} * {:2} = {:3}".format(n, i, result), end= " | ")
        n = n + 1
    print()
    i = i + 1 
````

## Problem 6

Write code to count the digits of a given number n.

### Pseudocode

```text
input n
count = 0
while n > 0
    remove the last digit from n
    count = count + 1
end while 
output count
```

````python
def digits(n):
    count = 0 
    while n > 0: 
        n = n // 10
        count += 1
    return count

print(digits(52632))
````

## Problem 7

Write code to print all prime numbers between any given pair of numbers n and m to the console.

### Pseudocode

```text
function prime_numbers 
    for each number in range from n to m
        for each divisor from 2 to number sqrt
            if number divisible by divisor 
                break
            end if
        else
            output number 
        end for
    end for
end function
```

````python
def prime_numbers(n, m):
    for number in range(n, m + 1):
        for divisor in range(2, int(number ** 0.5) + 1):
            if number % divisor == 0:
                break
        else:
            print(number)

prime_numbers(25, 50)
print(int(17**(1/2)))
````

# Problem 8

Write code that reverses a given string numbers. 

### Pseudocode

#### 1. string
```text
function reverse_string 
    revers = ""
    while word is not empty
        add last letter in revers
        remove last from word
    end while
    return revers
end function 
```
#### 2. integer 
```text 
function reverse_integer
    revers = 0
    while number is not 0
        digit = number mod 10
        reversed = reversed * 10 + digit 
        number = number div 10
    end while
    return revers
end function
```


````python
#string
def reverse_string(word):
    revers = ""
    while word != "":
        revers = revers + (word[-1])
        word = word[:-1]
    
    return revers

print(reverse_string("stressed"))
````

````python
#integer
def reverse_integer(number):
    revers = 0 
    while number != 0: 
        digit = number % 10
        revers = revers * 10 + digit
        number = number // 10
    
    return revers

print(reverse_integer(987654321))
````
